```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
%matplotlib inline
```

```python
df=pd.read_csv(r"C:\Users\hp\.cache\kagglehub\datasets\yasserh\student-marks-dataset\versions\1\Student_Marks.csv")
```

```python
df.head()
```

```python
df.drop(('number_courses'),axis=1,inplace=True)
```

```python
df.head()
```

```python
plt.scatter(df['time_study'],df['Marks'],c='r')
plt.xlabel('Time_study')
plt.ylabel('marks')
plt.show()
```

```python
df.corr()
```

```python
X=df[['time_study']]
y=df[['Marks']]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.20)
```

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
```

```python
regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
print('coefficient orr slope:', regressor.coef_)
print('intercept:', regressor.intercept_)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mse=mean_squared_error(y_pred,y_test)
mae=mean_absolute_error(y_pred,y_test)
score=r2_score(y_pred,y_test)
```

```python
print("MSE:",mse)
print("R2 Score:",score)
```

```python
plt.scatter(X,y)
plt.plot(X,regressor.predict(X))
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```
